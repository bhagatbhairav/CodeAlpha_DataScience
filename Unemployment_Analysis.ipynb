```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Year': [2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'Unemployment_Rate': [5.4, 5.6, 5.3, 5.4, 8.1, 6.7, 6.2]
}

df = pd.DataFrame(data)
print(df)
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['Year'], df['Unemployment_Rate'],
         color='blue', marker='o', linewidth=2)
plt.title('Unemployment Rate Analysis (2016-2022)')
plt.xlabel('Year')
plt.ylabel('Unemployment Rate (%)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df['Year'], df['Unemployment_Rate'],
        color=['green','green','green',
               'green','red','orange','yellow'])
plt.title('Unemployment Rate by Year')
plt.xlabel('Year')
plt.ylabel('Unemployment Rate (%)')
plt.grid(True)
plt.show()
```

```python
print("=== UNEMPLOYMENT ANALYSIS ===")
print()
print("Average Rate:", round(df['Unemployment_Rate'].mean(),2))
print("Highest Rate:", df['Unemployment_Rate'].max(),
      "- Year:", df.loc[df['Unemployment_Rate'].idxmax(),'Year'])
print("Lowest Rate:", df['Unemployment_Rate'].min(),
      "- Year:", df.loc[df['Unemployment_Rate'].idxmin(),'Year'])
print()
print("COVID-19 Impact:")
print("2019 Rate: 5.4%")
print("2020 Rate: 8.1%")
print("Increase:", round(8.1-5.4,1),"%")
```

```python
labels = ['Pre-COVID (2019)', 'COVID (2020)', 'Post-COVID (2021)']
values = [5.4, 8.1, 6.7]
colors = ['green', 'red', 'orange']

plt.figure(figsize=(8,5))
plt.bar(labels, values, color=colors)
plt.title('COVID-19 Impact on Unemployment')
plt.ylabel('Unemployment Rate (%)')
plt.grid(True)
plt.show()
```
