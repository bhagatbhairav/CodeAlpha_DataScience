```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
data = {
    'TV_Ad_Budget': [230, 44, 17, 152, 181,
                     263, 226, 120, 199, 66,
                     218, 240, 57, 145, 100],
    'Radio_Ad_Budget': [37, 39, 45, 41, 10,
                        25, 30, 19, 2, 23,
                        27, 13, 32, 35, 28],
    'Sales': [22, 10, 9, 18, 13,
              25, 20, 15, 11, 12,
              19, 17, 11, 16, 14]
}

df = pd.DataFrame(data)
print(df)
print()
print("Dataset Shape:", df.shape)
print("Average Sales:", df['Sales'].mean())
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['TV_Ad_Budget'],
            df['Sales'],
            color='blue',
            s=100)
plt.title('TV Ad Budget vs Sales')
plt.xlabel('TV Ad Budget')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
X = df[['TV_Ad_Budget']]
y = df['Sales']

model = LinearRegression()
model.fit(X, y)

print("=== SALES PREDICTION MODEL ===")
print()
print("Model Score (Accuracy):",
      round(model.score(X, y)*100, 2), "%")
print()
print("If TV Budget = 200:")
pred = model.predict([[200]])
print("Predicted Sales:", round(pred[0], 2))
print()
print("If TV Budget = 300:")
pred2 = model.predict([[300]])
print("Predicted Sales:", round(pred2[0], 2))
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['TV_Ad_Budget'],
            df['Sales'],
            color='blue',
            label='Actual Sales',
            s=100)

x_line = np.linspace(0, 300, 100)
y_line = model.predict(x_line.reshape(-1,1))
plt.plot(x_line, y_line,
         color='red',
         label='Prediction Line')

plt.title('Sales Prediction Model')
plt.xlabel('TV Ad Budget')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("=== SALES PREDICTION SUMMARY ===")
print()
print("Total Records:", len(df))
print("Average TV Budget:",
      round(df['TV_Ad_Budget'].mean(),2))
print("Average Sales:",
      round(df['Sales'].mean(),2))
print("Max Sales:", df['Sales'].max())
print("Min Sales:", df['Sales'].min())
print()
print("Key Insight:")
print("Higher TV Ad Budget = Higher Sales!")
print("Model Accuracy: 66.3%")
```
